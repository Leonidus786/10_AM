## Logical Operators
Logical operators are used to combine conditional statements.

```python
x = 20
```

```python
x < 25 and x > 10
```

```python
x = 5

print(x < 5 or x > 10)
```

```python
bool(x)
```

```python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(""))
print(bool(()))
print(bool([]))
print(bool({}))
```

```python
not x
```

## Membership Operators
Membership operators are used to test if a sequence is presented in an object

```python
x = ["apple", "banana"]

print("banana" in x)
```

```python
x = ["apple", "banana"]

print("pineapple" not in x)
```

## Bitwise Operators
Bitwise operators are used to compare (binary) numbers

```python
12 & 6
```

```python
12 | 6
```

```python
12 ^ 6
```
