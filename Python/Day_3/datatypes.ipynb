## Why datatypes ?

###  Understanding datatypes is essential because they define the identity and behavior of your data, directly determining which operations are valid and how the computer handles information.

## Primitive data types :- 

1. Integer - int()
2. Float - float()
3. Booleans - bool()
4. Strings - str()

# Integers -- Collection of whole numbers along with negative of natural numbers forms a collection of integers. ex - -5, -10, 23, 24, 30 3tc.

# Float -- Decimal wale numbers or we can say fractional numbers. ex- 1.3, -5.6, 7.8 etc

# Boolean -- True and False

# Strings -- Set of characters -- A-Z, a-z, Name etc.

```python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(""))
print(bool(()))
print(bool([]))
print(bool({}))
```
